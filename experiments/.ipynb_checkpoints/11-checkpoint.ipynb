```python
import matplotlib.pyplot as plt
import numpy as np

# ------------------------------
# Введите ваши данные здесь
# ------------------------------

# Модели
models = ['YOLOv11s', 'RT-DETR-L']

# Метрики
metrics = ['mAP@0.5', 'mAP@0.5:0.95']

# Средние значения для каждой модели и метрики
# Индексы: строки – модели, столбцы – метрики
means = np.array([
    [0.924, 0.687],   # YOLOv11s
    [0.930, 0.695]    # RT-DETR-L (пример, замените на свои числа)
])

# Полуширина доверительного интервала (margin of error) для 95% ДИ
# Для YOLOv11s вы можете рассчитать по своим 5 фолдам (пример ниже)
errors = np.array([
    [0.007, 0.007],   # YOLOv11s (полуширина ДИ)
    [0.018, 0.015]    # RT-DETR-L (пример, замените на свои)
])

# Если у вас есть только границы интервалов (low, high), то:
# margin = (high - low) / 2

# ------------------------------
# Построение графика
# ------------------------------

x = np.arange(len(metrics))          # позиции метрик
width = 0.35                          # ширина столбца

fig, ax = plt.subplots(figsize=(8, 6))

# Столбцы для каждой модели
bars1 = ax.bar(x - width/2, means[0], width, yerr=errors[0],
               capsize=5, label=models[0], color='#4C72B0', edgecolor='black')
bars2 = ax.bar(x + width/2, means[1], width, yerr=errors[1],
               capsize=5, label=models[1], color='#DD8452', edgecolor='black')

# Настройка внешнего вида
ax.set_ylabel('Значение метрики', fontsize=12)
ax.set_title('Сравнение метрик моделей с 95% доверительными интервалами', fontsize=14)
ax.set_xticks(x)
ax.set_xticklabels(metrics, fontsize=11)
ax.legend(fontsize=11)
ax.grid(axis='y', linestyle='--', alpha=0.6)

# Подписи значений над столбцами (среднее ± полуширина)
for bars, model_means, model_errors in zip([bars1, bars2], means, errors):
    for bar, mean_val, err_val in zip(bars, model_means, model_errors):
        height = bar.get_height()
        ax.text(bar.get_x() + bar.get_width()/2, height + err_val + 0.005,
                f'{mean_val:.3f}±{err_val:.3f}', ha='center', va='bottom', fontsize=8)

plt.tight_layout()
plt.show()
```
